```python
# ==========================================================
# AIML Experiment 4
# Binary Classification using Logistic Regression
# Performance Evaluation using Confusion Matrix
# ==========================================================

# Step 1: Clone GitHub Repository (Replace with your repository URL)
!git clone https://github.com/username/AIML-Lab-Experiment-4.git

# Step 2: Move into Repository
%cd AIML-Lab-Experiment-4

# Step 3: Import Required Libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Step 4: Load Dataset
dataset = pd.read_csv("bill_authentication.csv")

# Display first five rows
print("First Five Records")
print(dataset.head())

# Display dataset information
print("\nDataset Information")
print(dataset.info())

# Display statistical summary
print("\nDataset Summary")
print(dataset.describe())

# Step 5: Separate Features and Target
# Replace 'target' with your actual target column name
X = dataset.drop("target", axis=1)
y = dataset["target"]

# Step 6: Split Dataset into Training and Testing Sets
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

# Step 7: Train Logistic Regression Model
from sklearn.linear_model import LogisticRegression

model = LogisticRegression(max_iter=1000)

model.fit(X_train, y_train)

# Step 8: Predict Test Data
y_pred = model.predict(X_test)

# Step 9: Display Actual vs Predicted Values
results = pd.DataFrame({
    'Actual': y_test,
    'Predicted': y_pred
})

print("\nActual vs Predicted")
print(results.head(10))

# Step 10: Generate Confusion Matrix
from sklearn.metrics import confusion_matrix

cm = confusion_matrix(y_test, y_pred)

print("\nConfusion Matrix")
print(cm)

# Step 11: Plot Confusion Matrix
plt.figure(figsize=(6,5))

sns.heatmap(
    cm,
    annot=True,
    fmt='d',
    cmap='Blues',
    xticklabels=['Negative','Positive'],
    yticklabels=['Negative','Positive']
)

plt.xlabel("Predicted Class")
plt.ylabel("Actual Class")
plt.title("Confusion Matrix")
plt.show()

# Step 12: Calculate Accuracy
from sklearn.metrics import accuracy_score

accuracy = accuracy_score(y_test, y_pred)

print("\nAccuracy =", accuracy)

# Step 13: Classification Report
from sklearn.metrics import classification_report

print("\nClassification Report")
print(classification_report(y_test, y_pred))

# Step 14: Calculate Precision, Recall and F1 Score
from sklearn.metrics import precision_score
from sklearn.metrics import recall_score
from sklearn.metrics import f1_score

precision = precision_score(y_test, y_pred)
recall = recall_score(y_test, y_pred)
f1 = f1_score(y_test, y_pred)

print("\nPrecision =", precision)
print("Recall =", recall)
print("F1 Score =", f1)

# Step 15: Save Changes (Optional)
# Uncomment these commands if you want to commit changes from Colab

# !git add .
# !git commit -m "Completed AIML Experiment 4"
# !git push
```
